```python
import numpy as np
import tensorflow as tf
import cv2
import os

# Paths
frames_output_path = '/content/drive/MyDrive/computer vision/object_detection/video_frames'
output_frames_path = '/content/drive/MyDrive/computer vision/object_detection/video_frames_final'
os.makedirs(output_frames_path, exist_ok=True)

# Load the trained model
model = tf.keras.models.load_model('/content/drive/MyDrive/computer vision/object_detection/mobilenet_model.h5')

# Define your class labels (update this dictionary as per your dataset)
annotations_dict = {0: 'book', 1: 'phone'}

# Function to preprocess a single frame
def preprocess_frame(frame_path):
    img = tf.io.read_file(frame_path)
    img = tf.image.decode_jpeg(img, channels=3)
    img = tf.image.resize(img, (224, 224))  # Resize to MobileNet input size
    img = img / 255.0  # Normalize to [0, 1]
    return tf.expand_dims(img, axis=0)  # Add batch dimension

# Function to decode predictions
def decode_predictions(predictions, input_size=(224, 224)):
    # Assume predictions have the format: [[x, y, w, h, score, class_id], ...]
    decoded = []
    for pred in predictions[0]:
        x, y, w, h, score, class_id = pred
        if score > 0.5:  # Only keep predictions with a confidence score > 0.5
            bbox = [x * input_size[1], y * input_size[0], w * input_size[1], h * input_size[0]]
            decoded.append({'class_id': int(class_id), 'score': float(score), 'bbox': bbox})
    return decoded

# Function to draw predictions
def draw_predictions(frame, predictions, threshold=0.5):
    height, width, _ = frame.shape
    for pred in predictions:
        class_id, score, bbox = pred['class_id'], pred['score'], pred['bbox']
        if score > threshold:
            x, y, w, h = [int(dim) for dim in bbox]
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
            label = f"{annotations_dict[class_id]}: {score:.2f}"
            cv2.putText(frame, label, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
    return frame

# Process each frame
frame_files = sorted(os.listdir(frames_output_path))
for frame_file in frame_files:
    frame_path = os.path.join(frames_output_path, frame_file)

    # Load and preprocess the frame
    preprocessed_frame = preprocess_frame(frame_path)

    # Perform object detection
    predictions = model.predict(preprocessed_frame)

    # Post-process predictions
    decoded_predictions = decode_predictions(predictions)  # Decode model output

    # Draw predictions on the original frame
    original_frame = cv2.imread(frame_path)
    output_frame = draw_predictions(original_frame, decoded_predictions)

    # Save the output frame
    output_frame_path = os.path.join(output_frames_path, frame_file)
    cv2.imwrite(output_frame_path, output_frame)

print(f"Processed frames saved to {output_frames_path}")
```

```python
import cv2
import numpy as np

# Function to preprocess a single frame
def preprocess_frame(frame):
    img = cv2.resize(frame, (224, 224))  # Resize to MobileNet input size
    img = img / 255.0  # Normalize to [0, 1]
    return np.expand_dims(img, axis=0)  # Add batch dimension

# Function to draw bounding boxes
def draw_predictions(frame, predictions, threshold=0.5):
    for pred in predictions:
        class_id, score, bbox = pred['class_id'], pred['score'], pred['bbox']
        if score > threshold:
            x, y, w, h = bbox
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
            label = f"{annotations_dict[class_id]}: {score:.2f}"
            cv2.putText(frame, label, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
    return frame

# Perform inference on video
def detect_objects_in_video(video_path, model, output_path):
    cap = cv2.VideoCapture(video_path)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = int(cap.get(cv2.CAP_PROP_FPS))

    # Define the codec and create a VideoWriter object
    out = cv2.VideoWriter(output_path, cv2.VideoWriter_fourcc(*'XVID'), fps, (width, height))

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        # Preprocess frame and make predictions
        preprocessed_frame = preprocess_frame(frame)
        predictions = model.predict(preprocessed_frame)  # Use trained model for prediction

        # Post-process predictions and draw on frame
        output_frame = draw_predictions(frame, predictions)

        # Write the frame to the output video
        out.write(output_frame)

    cap.release()
    out.release()
    print("Video saved to:", output_path)

# Example usage
detect_objects_in_video(
    video_path='/content/drive/MyDrive/computer vision/Dataset2.avi',
    model=tf.keras.models.load_model('/content/drive/MyDrive/computer vision/object_detection/mobilenet_model.h5'),
    output_path='/content/drive/MyDrive/computer vision/object_detection/'
)
```

```python
def load_data(image_dir, annotation_file):
    with open(annotation_file) as f:
        annotations = json.load(f)

    images = []
    bboxes = []
    labels = []

    for annotation in annotations['annotations']:
        image_id = annotation['image_id']
        bbox = annotation['bbox']
        category_id = annotation['category_id']

        image_filename = f"frame_{image_id}.jpg"
        image_path = os.path.join(image_dir, image_filename)

        if os.path.exists(image_path):  # Check if image file exists
            image = load_img(image_path, target_size=(224, 224))  # Resize image to 224x224 for MobileNet
            image = img_to_array(image)
            images.append(image)

            bboxes.append(bbox)
            labels.append(category_id)  # 1 for book, 2 for phone
        else:
            print(f"Warning: {image_filename} not found.")

    # Convert to numpy arrays after the loop
    images = np.array(images)
    bboxes = np.array(bboxes)
    labels = np.array(labels)

    return images, bboxes, labels

# Example usage
image_dir = '/content/drive/MyDrive/computer vision/object_detection/images'
annotation_file = '/content/drive/MyDrive/computer vision/object_detection/annotation.json'

images, bboxes, labels = load_data(image_dir, annotation_file)
```
